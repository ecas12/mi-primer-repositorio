En esta primera celda se ulilizaron las librerias y el TensorFlow y Keras para construirla red neuronal multicapa,las utilidades de Scikit-Learn para importar el conjunto de datos de las flores.

```python
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Dense
from sklearn.datasets import load_iris
from sklearn.model_selection import train_test_split
from tensorflow.keras.utils import to_categorical
```

Celda 2
Aquí realizo la garga del dataset clásico de Iris.Como es un proplema  de clasificación con tres tipos diferentes de flores,serealizo las etiquetas numéricas a forma one-hot encoging para que la red pueda interpretarlas correctamente en la salida.Después separé la información dejando un 80% para entrenar el modelo y un 20% para ponerlo a prueba.

```python
# Cargar y preparar el conjunto de datos de Iris
iris = load_iris()
X = iris.data
y = iris.target

# Convertir las etiquetas a one-hot encoding para clasificación multiclase
y_encoded = to_categorical(y)

# Dividir en datos de entrenamiento y prueba
X_train, X_test, y_train, y_test = train_test_split(X, y_encoded, test_size=0.2, random_state=42)
```

Celda 3
En esta celda se diseñola arquitectura de la red neuronal utilizado con modelo secuencia(Sequential).Definí una capa aculta de 10 neuronas con función de activación ReLu para aprender los patrones de los datos,y una capa de salida con 3 neuronas y función Softmax,que es la ideal para darnos las probabilidades de cada una de las tres clases de floresl.

```python
# Construir el modelo de red neuronal multicapa (Perceptrón Multicapa)
model = Sequential()
model.add(Dense(10, input_dim=4, activation='relu')) # Capa oculta con 10 neuronas
model.add(Dense(3, activation='softmax'))           # Capa de salida con 3 neuronas (clases de iris)

# Compilar el modelo
model.compile(loss='categorical_crossentropy', optimizer='adam', metrics=['accuracy'])
```

Celda 4
Procedí a entrenar la red neuronal pasándole los datos de entrenamiento(x_train y y_train).Configuramos un ciclo de 50 épocas(epochs)para que el modelo vaya ajustando sus pesos de manera iterativa y aprenda a clacificar las flores con mayor preción.

```python
# Entrenar el modelo
history = model.fit(X_train, y_train, epochs=50, batch_size=5, verbose=1)
```

Celda 5
Por último,evalúe el desempeñodel modelo utilizado el conjunto de prueba(x_test y y_test) que la red no había visto antes.Esto me permitió comprobar el porcentaje de acierto real que obtuvo la red neuronal al clasificar las flores deIris.


```python
# Evaluar el modelo
loss, accuracy = model.evaluate(X_test, y_test)
print(f"Precisión en el conjunto de prueba: {accuracy * 100:.2f}%")
```
