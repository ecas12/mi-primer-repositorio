```python
import numpy as np

def neurona_not(x):
    # Al ser una operacion inversa, usamos un peso negativo
    w = -1
    theta = 0 # umbral

    # se calcula la suma ponderada
    suma_ponderada = w * x # Calculate suma_ponderada

    if suma_ponderada >= theta:
        return 1
    else: # Corrected from 'elsa:'
        return 0

# comprobamos con la tabla de verdad
entrada = [0, 1]

print("tabla de verdad de la neurona NOT:")
print("x \t y")
for x in entrada:
    y = neurona_not(x)
    print(f"{x} \t {y}")
```

Explicacion de la neurona NOT:
La neurona NOT tiene el objetivo de invertir el valor que recibe en su unica entrada.Para logar esta inversion matematica mediante el modelo de McCulloch-Pitts,le configuramos un peso negativo de -1 y establecemos el umbral en 0. Si el valor ingresado  es 1,el resultado de la multiplicacion es -1,quedando por debajo de cero e impidiendo que la neurona (regresa 0).

```python
import numpy as np
def neurona_or(x1,x2):
  w1 = 1
  w2 = 1
  theta = 1 #umbral
  suma_ponderada = (x1 * w1) + (x2 * w2) # Corrected variable name

  if suma_ponderada >= theta: # Corrected condition and indentation
      return 1
  else: # Corrected from 'elsa:' and indentation
      return 0
```

Explicacion de la neurona OR:
Esta neurona procesa dos entradas distintas asignandole a cada  una un peso 1 y fijamos el umbral de activacion  en 1. Cuando ambas  entradas  valen 0,la suma es 0,por lo que no supera el umbral y la salida es 0. En cambio,si al menos una de las entradas es 1,la suma total llega a 1 o 2,igualando o superando el umbral. Esto hace que la neurona  se active y devuelva un 1.
