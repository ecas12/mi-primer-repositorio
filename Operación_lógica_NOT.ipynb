Operación lógica NOT con una neurona artificial

```python
# operación lógica NOT con una neurona artificial
def escalon(valor):
  if valor >= 0:
    return 1
  else:
    return 0
# Parámetro de la neurona
peso = -1
sesgo = 0.5
#entrada pocibles
entradas = [0,1]
print("Entrada | salida NOT")
for x in entradas:
  suma = (x * peso) + sesgo
  salida = escalon(suma)
  print(f"{x} | {salida}")
```

En este ejercicio utilicé una neurona artificial para representar la operación lógica NOT.A diferencia de OR,esta neurona solamente recibe una entrada.
La neurona multiplica la entrada por un peso negativo y suma el sesgo.Después la función escalón determina el resultado.Cuando la entrada es uno,la salida es cero.
Con este ejemplo se observa como una neurona puede representar una operación lógica mediante sus parámetros y una función de activación.
