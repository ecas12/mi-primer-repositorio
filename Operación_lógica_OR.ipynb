Operación lógica OR con una neurona artificial

```python
# operación lógica OR con una neurona artificial
def escalon(valor):
  if valor >= 0:
    return 1
  else:
    return 0

# Entradas y parámetros de la neurona
peso1 = 1
peso2 = 1
sesgo = -0.5

# Combinación de entradas (corregido 'o' por '0' y la tupla inicial)
entradas = [(0, 0), (0, 1), (1, 0), (1, 1)]

print("Entrada 1 | Entrada 2 | Salida OR")
for x1, x2 in entradas:
  # Combinación lineal con ambos pesos y el sesgo
  suma = x1 * peso1 + x2 * peso2 + sesgo
  salida = escalon(suma)
  print(f"    {x1}     |     {x2}     |    {salida}")
```

En este ejercicio utilícé una neurona artificial para reprecentar la operación lógica NOT. A diferencia de OR, esta neurona solamente recibe una entrada.
La neurona se multiplica la entrada por un peso negativo y suma el sesgo.Después,la función escalón determina el resultado.Cuando la entrada es cero,la salida es uno;cuando la entrada es uno,la salida es cero.
En este ejemplo se observa cómo una neurona puede representar una operación lógica mediante sus parámetros y una funcion de activación.
